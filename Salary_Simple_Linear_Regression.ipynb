# Imports

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

# Loading Dataset from Kaggle

```python
#Importing Dataset from Kaggle - "Salary data - Simple linear regression"

#Import pandas
import pandas as pd

df = pd.read_csv('Salary_Data.csv')

X = df[["YearsExperience"]]
y = df["Salary"]

print(df.head())
```

# Train/ Test split

This block spits the data set into training rows which the model learns from and test rows which the model never learns from. We can use them to check the performance on "new" data.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    test_size=0.20, # 20%test, 80% train
    random_state=1 # same split every time
)
```

# Train a Linear Regression Model
This block creates a linear regression model and trains it. Training means learning the best slope amd intercept from the training data

```python
model = LinearRegression() # make the model
model.fit(X_train, y_train) #train (learn slope + intercept)

print ("slope(m): ", model.coef_[0]) # weight for x
print ("intercept(b): ", model.intercept_)
```

# Predict and Measure Error (MSE)

This block uses the trained model to make predictions on the test set then computse MSE (mean squared error). Lower MSE means the perdiction are closer to the true values on average.

```python
y_pred = model.predict(X_test) #predictions (y-hat)
mse = mean_squared_error(y_test,y_pred) #average squared mistake

print("test MSE: ", mse)

#Examples

for x_val, true_y, pred_y in zip(
    X_test.to_numpy().flatten(), y_test, y_pred
):
    print(
        "Years of experience =", x_val,
        "| Actual salary =", true_y,
        "| Predicted salary =", round(float(pred_y), 2)
    )
```
